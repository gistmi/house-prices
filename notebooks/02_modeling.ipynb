```python
%load_ext autoreload
%autoreload 2

import sys
sys.path.append('..')

import numpy as np
import pandas as pd

from src.cleaning import fit_fill_values, clean
from src.features import make_features
```

```python
raw_train = pd.read_csv('../data/raw/train.csv')
raw_test = pd.read_csv('../data/raw/test.csv')

mask = (raw_train['GrLivArea'] > 4000) & (raw_train['SalePrice'] < 300000)
raw_train = raw_train[~mask].reset_index(drop=True)

fv = fit_fill_values(raw_train)
train = clean(raw_train, fv)
test = clean(raw_test, fv)

y = np.log1p(train['SalePrice'])
print(train.shape, test.shape)
print(train.isnull().sum().sum(), test.isnull().sum().sum())
```

```python
train_f = make_features(train)
test_f = make_features(test)

n_train = len(train_f)
all_X = pd.concat(
    [train_f.drop(columns=['Id', 'SalePrice']), test_f.drop(columns=['Id'])],
    ignore_index=True
)
all_X = pd.get_dummies(all_X, dtype=int)

X = all_X.iloc[:n_train]
X_test = all_X.iloc[n_train:].reset_index(drop=True)

print(X.shape, X_test.shape)
print(X.isnull().sum().sum(), X.dtypes.unique())
```
